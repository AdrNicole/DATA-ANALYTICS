#Ejercicio 1: Suma de números pares
'Primero ingreso los numero con la condicion de que sean enteros. Segundo, coloco un contador llamado suma que acumula los numeros entre el rango A y B de modo que cada numero i se añade al contador si y solo si es par es decir si su resto es 0 al dividirlo entre 2, de lo contrario (si no es par) entonces suma 0, despues hay una restriccion en caso la persona ingrese el primer numero mayor al segundo numero (B>A) entonces el limite inferior será B y el limite superior A'

```python
A = int(input("Ingresar primer numero entero: "))
B = int(input("Ingresar segundo numero entero: "))

suma=0 # condicion

if A<=B: # condicion
  for i in range(A,B):
    if i%2==0: # es par?
      suma+=i # si es par
    else:
      suma+=0 # no es par
else:
  for i in range(B,A): # 2da condicion
    if i%2==0: # es par?
      suma+=i # si es par

print(f'La suma de numeros pares es: {suma}')
```

#Ejercicio 2: Calculadora de áreas

'Se coloca un ingreso general del nombre de la figura con la condicion que sean letras. Luego aplico de condiciones una para cada figura y en cada caso se ingresan las variables respectivas para guardar en la variable area el calculo de esta y finalmente se imprime el area con print. Por ultimo se añade una condicion sino se ingresa ningun nombre de la figura que vuelva a ingresar el nombre'

```python
import math

figura = str(input('Ingrese figura (cuadrado, triangulo o circulo): '))

if figura == 'cuadrado':
  l = float(input('Ingrese largo: '))
  print(f'El area es {l**2} m2')

elif figura == 'triangulo':
  b = float(input('Ingrese base: '))
  h = float(input('Ingrese altura: '))
  area = (b*h)/2
  print(f'El area es {area} m2')

elif figura == 'circulo':
  r = float(input('Ingrese radio: '))
  area = math.pi*r**2
  print(f'El area es {area} m2')

else:
  print('Vuelve a ingresar el nombre de la figura: ')
```

#Ejercicio 3: Adivina el número
'Importamos la libreria random y colocamos un input para permitir al usuario ingresar un numero a comparar con el numero aleatorio r. Luego con el bucle while true añadimos 3 condiciones de las cuales debe ir primero la igualdad de ambos numeros para que imprima correctamente el texto de felicidades. Seguido de este  colocamos la condicion de numero ingresado mayor al numero random y pedimos al usuario volver a ingresar el numero y guardamos la variable para que el bucle se encargue de volver a compararlos ya que renombramos la variable en cada caso con el nuevo numero ingresado'

```python
import random
r=random.randint(1,100)
print('------Bienvenido querido usuario a ¡Adivina Adivinador!------')
N=int(input('Ingresar numero entre 1 y 100: '))
#print(r)

while True:
  if r==N:
    print(f'Felicidades ¡ADIVINASTE! {r} es igual a {N}')
    break

  elif N<r:
    print('El numero ingresado es menor👀')
    n1=int(input('Vuelve a ingresar: '))
    N=n1

  elif N>r:
    print('El numero ingresado es mayor👀')
    n2=int(input('Vuelve a ingresar: '))
    N=n2
```

#Ejercicio 4: Carrera de pollos
'Primero se muestra la entrada de la velocidad metros/segundos válida y con válida me refiero a que sea multiplo de 100 para que el contador de los segundos no se vea alterado en cada iteracion. Luego se inscriben el inicio del pollo=0, el contador de segundos=0 y el acumulador de las lineas de carrera=[].

Ahora si se entra de lleno al bucle while, mientras el recorrido del pollo sea diferente a la distancia meta seguirá iterando la condicion. La cual consiste en validar que el numero ingresado sea multiplo y que el recorrido del pollo sea menor igual a la distancia meta.

Dentro de la condicion se coloca el contador de segundos y el contador del recorrido del pollo y el acumulador de la linea de 21 giones bajos a la lista 'a' y posteriormente se juntan los guiones de la lista (sin la separacion de comas) mediante join y se añade a la ultima posicion del string b[:-1] un pollito 🐥 con slicing.

Finalmente se añade la condicional else si no cumple y con un comentario para que el usuario vuelva a ingresar una distancia junto a un break para cortar la iteracion. Se imprime el resultado final de los segundos recorridos por el pollo'

psd: abajo estan las consolas que se usanron para crear el recorrido del 🐥. ¡Divide y vencerás!

```python

print('Bienvenido a ¡CARRERA EMPLUMADA!')
M = int(input('Ingrese velocidad en metros/segundos: ')) #20 #25 #50 #5 #10

pollo = 0
segundos = 0
meta = 100
a= [] # acumulador de linea de carrera

while pollo!=meta:
  if pollo<=meta and meta%M==0: # si M (20) es multiplo de meta(100) o si pollo<meta
    segundos+=1 # contador de segundos 🐥
    pollo+=M
    a.append('_____________________')
    b=''.join(a)
    b=b[:-1]+('🐥')
    print(b)
  else:
    print('Ingresar una velocidad válida')
    break

if pollo==meta:
  print(f'El pollito ha llegado a la meta en {segundos} segundos')
```
