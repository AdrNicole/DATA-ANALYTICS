#Ejercicio 2
Fabricación de cerveza artesanal

El objetivo del ejercicio es calcular la media, la desviación estándar, el LIC y el LSC del
porcentaje de alcohol y guardarlos en variables.
Para asegurarte de que los lotes de cerveza se estén elaborando de forma consistente y que el
nivel de alcohol sea el adecuado, una vez calculadas las medidas anteriores, indicar cuáles
valores quedarían fuera de los límites de control y cuántos son.


```python
import numpy as np
import pandas as pd
#nombres=np.loadtxt("alcohol_cerveza.csv",delimiter=',',dtype=str,encoding='utf-8')
#ruta = '/alcohol_cerveza.csv'
#df = pd.read_csv(ruta)

array = np.loadtxt('/alcohol_cerveza.csv')
print(array)

media = np.mean(array)
desviacion =np.std(array)
lic = media - (3*desviacion)
lsc = media + (3*desviacion)
condicion1 = array< lic
condicion2 = array>lsc
valores_fuera = array[np.logical_or(condicion1,condicion2)]
#fuera_de_control = quality_column[np.logical_or(quality_column < lic, quality_column > lsc)]

print(f'la media es: {media}')
print(f'desviacion es: {desviacion}')
print(f'el límite inferior de control (LIC): {lic}')
print(f'límite superior de control (LSC): {lsc}')
print('Valores fuera de control:')
print(valores_fuera if valores_fuera.size > 0 else "No se encontraron valores fuera de control.")
```

De lo anterior se muestra que los datos deberian estar en el rango que va de 3.957 a 6.053 unidades porcentuales de alcohol de cada lote. Dados los resultados hay dos valores atípicos extremos. Ambos se encuentran por debajo del limite inferior de control x1=0 y x2=3.87 lo que afecta el promedio que fue de 5.005 unidades porcentuales. Es así, que se puede considerar
que el proceso está fuera de control y se requiere una acción correctiva para investigar y corregir la causa de la variación.
