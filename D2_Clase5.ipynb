# Ejercicio 1
Para desarrollar este desafío necesitarás los siguientes archivos:
`incidents.pkl`
`officers.pkl`
`subjects.pkl`

a. Carga los datos y crea un DataFrame con cada uno de ellos.

```python
import pickle
import pandas as pd
import numpy as np
```

```python
df1 = pd.read_pickle("incidents.pkl")
df2 = pd.read_pickle("officers.pkl")
df3 = pd.read_pickle("subjects.pkl")

df1, df2, df3
```

b. Genera una tabla que contenga la unión de las 3 tablas. hint: utiliza sufijos para para las columnas que se llaman igual usando el parámetro suffixes de pd.merge().

```python
# Fusionar los DataFrames usando case_number" como clave de combinación
df1_df2 = pd.merge(df1, df2, on ='case_number', how = 'inner')
len(df1_df2) # 370
df1_df2_df3 = pd.merge(df1_df2, df3, on ='case_number', how = 'inner', suffixes = ('_df','_df3'))
df1_df2_df3.head(1)
len(df1_df2_df3) # 377

df1_df2_df3.head()
```

c. Verifica si hay filas duplicadas; si es así, elimínalas.


```python
#from numpy._core.defchararray import count
df1_df2_df3_copia = df1_df2_df3.copy()
df1_df2_df3_copia.head(1)        # 377 rows con duplicados

# creo una nueva columna de boleanos siendo True = duplicados y False = No duplicado
df1_df2_df3_copia['duplicados'] = df1_df2_df3_copia.duplicated()
df1_df2_df3_copia[df1_df2_df3_copia.duplicated()]
len(df1_df2_df3_copia[df1_df2_df3_copia['duplicados']==False])  # como es 377 otra vez entonces no hubo duplicados

# a continuacion se muestran las filas sin duplicados
df1_df2_df3_copia[df1_df2_df3_copia['duplicados']==False].head()
```


d. ¿Cuántos sujetos de género F hay en el DataFrame resultante? hint: usa el método
.value_counts() sobre la columna.

```python
# sujetos -> df3
df1_df2_df3_copia['gender_df3'].value_counts()
# 9 sujetos con el genero F
```

e. ¿En cuántos números de caso hay por lo menos una sospechosa que sea mujer? hint:
utiliza el método unique() para obtener los valores únicos de una columna específica de un DataFrame luego de filtrar.

```python
df4 = pd.merge(df1,df2, on= 'case_number',how='inner')
df4_casos_gender = pd.concat([df4['case_number'], df4['gender']],axis=1)
# filtrar < extraer < contar
df4_casos_gender_F = df4_casos_gender[df4['gender']=='F']
df4_casos_gender_F_unico = df4_casos_gender_F['case_number'].unique()
resp = len(df4_casos_gender_F_unico)
print(f'hay {resp} casos donde por lo menos hay una sospechosa mujer')
```

f. Genera una tabla pivote que muestre en las filas el género del oficial y en las columnas el género del subject. ¿Cómo interpretas los valores que muestra esta vista?

A continuacion se muestra lo siguiente:
- las mujeres oficiales son las que atienden el menor numero de casos cometidos por sujetos mujeres (2 case_number).
- las mujeres oficiales son las que atienden poco mas de casos cometidos por hombres (18 case_number).
- los hombres oficiales son los que atienden más casos cometidos por sujetos hombres (350 case_number).
- los hombres oficales atienden menos casos cuando son cometidos por sujetos mujeres (7 case_number).

En general hay una mayor participacion de hombres tanto de oficiales que atienden el caso como los sujetos que los cometen; mientras que en el caso de las mujeres su participación es notablemente limitada en este ámbito.

```python
tabla_pivote = pd.pivot_table(
    df1_df2_df3_copia,
    values = 'case_number',
    index= 'gender_df',
    columns= 'gender_df3',
    aggfunc= 'count'
)

tabla_pivote
```

# Ejercicio 2
Para continuar con el desarrollo de este desafío, necesitarás el archivo
`Cleaned_DS_Jobs.csv`

a. Carga los datos y crea un DataFrame con ellos.

```python
df5 = pd.read_csv('Cleaned_DS_Jobs.csv')
df5
```

b. Utiliza la siguiente lista de valores que serán considerados como nulos:
["na", "NA",-1, "0", "-1", "null", "n/a", "N/A", "NULL"]
(hint: utiliza el método replace para reemplazar los valores indicados por np.nan)

```python
# Reemplazar inf y -inf por NaN manualmente
df5.replace(["na", "NA",-1, "0", "-1", "null", "n/a", "N/A", "NULL"], np.nan, inplace=True)
df5
```

c. Elimina todas las filas con datos faltantes. (hint: utiliza el método .dropna())

```python
# Copia para eliminar filas con valores faltantes
df_sin_filas_na = df5.copy().dropna()

df_sin_filas_na.head(3)
```


d. Apartir de la columna “Salary Estimate”, genera dos columnas: Salario Estimado Mínimo y Máximo. (hint: Utiliza el método apply sobre la columna.)

```python
df_sin_filas_na[['Salario_Estimado_Min','Salario_Estimado_Max']] = df_sin_filas_na['Salary Estimate'].apply(lambda w: pd.Series(str(w).split('-')))
df_sin_filas_na.head(3)
```

e. Realiza la recodificación de la columna Size con los valores de la siguiente tabla: (hint: utilice reemplazo con diccionario usando el método replace sobre la columna.)

```python
df_sin_filas_na.replace(['10000+ employees','5001 to 10000 employees','1001 to 5000 employees','201 to 500 employees','51 to 200 employees','501 to 1000 employees','Unknown'],['MegaEmpresas','Grandes Empresas','Medianas Empresas','Pequeñas Empresas','Pequeñas Grandes Empresas','Microempresas','Empresas sin Información'], inplace=True)
df_sin_filas_na.head()
```

f. Finalmente, genera una tabla pivote que muestre la media del salario estimado mínimo y
la media del salario estimado máximo por tamaño de empresa. (hint: utiliza
pd.pivot_table para generar la vista adecuada con las columnas generadas.)

```python
#(df_sin_filas_na['Salario_Estimado_Max'], df_sin_filas_na['Salario_Estimado_Min'])

df_sin_filas_na['Salario_Estimado_Minimo']= df_sin_filas_na['Salario_Estimado_Min'].fillna(0).apply(lambda w: int(w))
df_sin_filas_na['Salario_Estimado_Maximo']=df_sin_filas_na['Salario_Estimado_Max'].apply(lambda w: int(w))
df_sin_filas_na

tabla_pivote2 = pd.pivot_table(
    df_sin_filas_na,
    values = ['Salario_Estimado_Minimo','Salario_Estimado_Maximo'],
    index= 'Size',
    #columns= ,
    aggfunc= 'mean'
)
tabla_pivote2.round(2)
```
