# Desafío 4
Para desarrollar este desafío necesitarás el archivo

> `resumen_resultados_astronautas.csv`

El IMC es una medida importante que nos ayuda a determinar si una persona
tiene un peso saludable para poder ser un astronauta. Para calcularlo se utiliza la
fórmula `IMC = m / h*h`

Donde m corresponde al peso del astronauta en kilogramos, y h su estatura en metros
El rendimiento en las diferentes evaluaciones permitirá también aceptar o rechazar a los
postulantes.

1. Crea un DataFrame con el archivo entregado.




```python
import numpy as np
import pandas as pd

df = pd.read_csv('resumen_resultados_astronautas.csv')
df.head()
#df.info()
#df.describe() # genera medidas como count, mean, std, min, 25%, 50%, 75% y max
```

2. Crea la columna "IMC" usando la fórmula entregada.


```python
# corregimos la altura que está en centímetros, y digo esto porque ninguna persona tiene 180 metros de altura.
df['altura_corregida'] = df['altura']/100

# ahora si, calculo IMC
IMC = df['peso']/(df['altura_corregida']*df['altura_corregida'])
df
df['IMC'] = IMC
df.head()
```

3.1 Crea la columna "promedio_evaluaciones" con el promedio de las 4 columnas
de evaluaciones.

```python
# creacion de tabla evaluaciones
evaluaciones = df.iloc[:,4:8]
evaluaciones.head()
  # Forma1 (loc): sumar columnas por nombres
suma1 = evaluaciones['evaluacion_fisica']+ evaluaciones['evaluacion_psicologica']+ evaluaciones['evaluacion_adaptativa']+ evaluaciones['evaluacion_medica']
evaluaciones['suma1']=suma1
evaluaciones.head()
  # Forma2 (iloc): sumar columnas por su indice
evaluaciones.iloc[:3] # <- fila
evaluaciones.iloc[:,1] # <- columna
suma2 = evaluaciones.iloc[:,0] + evaluaciones.iloc[:,1] + evaluaciones.iloc[:,2] + evaluaciones.iloc[:,3]
evaluaciones['suma2']=suma2
evaluaciones.head()
  # incorporar la columna 'promedio= suma2/4' a la tabla principal df
df['promedio_evaluaciones'] = suma2/4
df.head()
```

3.1.   Si el IMC está entre 18.5 y 24.9, entonces el candidato
tendrá un IMC adecuado para ser astronauta. Filtra el DataFrame para
obtener solo aquellas filas donde el IMC esté dentro de los límites
entregados.

```python
df_IMC_Adecuado = df
IMC_adecuado= df_IMC_Adecuado[np.logical_and(df['IMC']>=18.5,df['IMC']<=24.9)]
#(df['IMC']>=18.5) & (df['IMC']<=24.9)
len(IMC_adecuado)
IMC_adecuado.tail(6) # verificamos, que la fila 26 en la tabla df es False y este filtro quita esa fila, por tanto este filtro es correcto
```

4. El promedio de las 4 evaluaciones debe ser mayor o igual a 87. Filtra el
DataFrame resultante y agrega esta condición al filtro aplicado en el paso
anterior.

```python
IMC_adecuado_evalu87 = IMC_adecuado[np.logical_or(IMC_adecuado['promedio_evaluaciones']>87, IMC_adecuado['promedio_evaluaciones']==87)]
IMC_adecuado_evalu87.head()
```

6. Modifica el valor de la columna "califica", para que diga "Si" en todas las filas del
DataFrame resultante.

```python
IMC_adecuado_evalu87['califica']='Si'
IMC_adecuado_evalu87
```
