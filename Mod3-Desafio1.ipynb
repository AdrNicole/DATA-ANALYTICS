```python
import pandas as pd
df = pd.read_csv('ds_salaries.csv')
df.head()
```

```python
# 1. El promedio general de los salarios, desviación estándar, quintiles y rango.
# (hint: escoge la columna adecuada para comparar)

media = df['salary'].mean()
desviacion_estandar = df['salary'].std()
q1 = df['salary'].quantile(0.2)
q5 = df['salary'].quantile(0.8)
minimo = df['salary'].min()
maximo = df['salary'].max()
rango = maximo-minimo

print(f'media: {round(media,4)} \n desviacion_estandar: {round(desviacion_estandar,4)}')
print(f'qintil 1: {round(q1,4)} \n quintil 5: {round(q5,4)}\n rango: {round(rango,4)}')
```

```python
# 2. Una comparación entre los salarios agrupados por 3 diferentes categorías a 
# tu elección. ¿Para cuál de ellas las medidas de tendencia central son más
# representativas? ¿Para cuál son menos representativas? Explica.

# primera forma
media_salario_por_employment_type = df.groupby('employment_type')['salary'].mean()
media_salario_por_company_size = df.groupby('company_size')['salary'].mean()
media_salario_por_experience_level = df.groupby('experience_level')['salary'].mean()
print(round(media_salario_por_employment_type,2),'\n')
print(round(media_salario_por_company_size,2),'\n')
print(round(media_salario_por_experience_level,2))

# Grupo1: El promedio de salarios de los empleados por su tipo de empleo 'trabajo freelance' (FL) es superior 
# comparado con las demás categorías 'tiempo completo' (FT), por 'contrato' (CT) 
# y 'tiempo parcial' (PT) que es signicativamente inferior.

# Grupo2: El promedio de salarios de los empleados por el tamaño de la compañia es significativamente más alto en una empresa grande 
# y con más de 250 empleados (L) que en las demás agrupaciones, además cabe mencionar que los salarios en una empresa mediana
# con 50 a 250 empleados (M) es bajo comparandolo con una empresa pequeña com menos de 50 empleados (S) 
# donde el salario medio es casi el doble.

# Grupo3: Finalmente, el promedio del salario por el nivel de experiencia presenta un ligero empate entre los salarios recibidos
# por empleados con nivel medio (MI) y los recibidos por empleado con un nivel ejecutivo / director (EX), mientras que los salarios medios 
# de los empleados del nivel senior / experto (SE) ganan ligeramente menos que los de nivel de entrada / junior (EN).

# la medida de tendencia central, promedio, es más representativa para los salarios por tamaño de compañia, pues refleja 
# las caracteristicas del grupo más grande, esto nos permite hacer inferencias confiables sobre toda la población.
# En contraparte la menos representativa es la agrupación por nivel de experiencia, pues se requiere más información que solo el
# promedio para afirmar que los salarios medios de los empleados del nivel senior / experto (SE) sean ligeramente menores 
# que los de nivel de entrada / junior (EN).
```

```python
# segunda forma
tabla_pivote = pd.pivot_table(df,
                              #columns = ,
                              index = ['employment_type','company_size','experience_level'],
                              values= 'salary',
                              aggfunc= 'mean')    
round(tabla_pivote,2)
```

```python
# 3. ¿Qué cargos reciben mejores sueldos en las empresas con sede en Estados Unidos?
# cargos: job_title
# sueldos: salary y salary_in_usd
# usa: company_location

TablaPivote2 = pd.pivot_table(df[df['company_location']=='US'],
                             index = 'job_title',
                             values = 'salary',
                             aggfunc = 'max')
TablaPivote3 = TablaPivote2.sort_values(by='salary', ascending=False)
TablaPivote3 = TablaPivote3.reset_index()
TablaPivote3.head(10)
```
