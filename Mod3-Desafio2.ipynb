```python
# a. pertenecer a África
conteo_Samerica = df['continent']=='southamerica'
conteo_Namerica = df['continent']=='northamerica'
conteo_europa = df['continent']=='europe'
conteo_africa = df['continent']=='africa'
conteo_asia = df['continent']=='asia'

total = conteo_Samerica.sum() + conteo_Namerica.sum() + conteo_europa.sum() + conteo_africa.sum() + conteo_asia.sum()
p_africa = conteo_africa.sum()/total
print(f'la probabilidad de que eso ocurra es {p_africa}')
```

```python
# b. pertenecer a Asia
def probabilidad_pais(tabla, columna, numerador):
    global total
    conteo_pais = tabla[columna].value_counts()[numerador]
    condicion = tabla[columna]==numerador
    p_pais = conteo_pais/total
    probabilidad = print(f'la probabilidad de que eso ocurra es {p_pais}')
    return conteo_pais
    
asia = probabilidad_pais(df,'continent','asia')
asia
```
