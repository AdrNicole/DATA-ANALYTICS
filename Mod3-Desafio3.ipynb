```python
# A. Calculamos los cuartiles 1 (25%) y 3 (75%) de la columna
Q1 = df2['Co2-Emissions'].quantile(0.25)
Q3 = df2['Co2-Emissions'].quantile(0.75)

# B. Calculamos el Rango Intercuartílico (IQR)
IQR = Q3 - Q1

# C. Definimos la frontera para los datos atípicos superiores
limite_superior = Q3 + (1.5 * IQR)

# D. Filtramos el DataFrame para obtener solo los países que superan ese límite
paises_atipicos = df2[df2['Co2-Emissions'] > limite_superior]

# E. Ordenamos de mayor a menor para ver el resultado impecable
paises_atipicos_ordenados = paises_atipicos[['Country', 'Co2-Emissions']].sort_values(by='Co2-Emissions', ascending=False)

# Mostramos cuántos países son atípicos y la lista
print(f"Se encontraron {len(paises_atipicos_ordenados)} países atípicos de los 195 totales.")
paises_atipicos_ordenados.head()
```

```python
lista = []
for i in df3:
    Q1 = df2[i].quantile(0.25)
    Q3 = df2[i].quantile(0.75)

    IQR = Q3 - Q1
    limite_superior= Q3 + (1.5* IQR)
    
    paises_atipicos = df2[df2[i] > limite_superior]
    paises_atipicos_ordenados = paises_atipicos[['Country',i]].sort_values(by=i,ascending=False)
    lista.append(paises_atipicos.index)
    
    print(f'Se encontraron {len(paises_atipicos_ordenados)} países atípicos de los 195 totales.')

    paises_sin_atipicos = df2[df2[i] <= limite_superior]
    paises_sin_atipicos_ordenados = paises_sin_atipicos[['Country',i]].sort_values(by=i,ascending=False)
    print(paises_sin_atipicos_ordenados)

print(lista)
```

```python
# aplanamos la lista con los indices de los valores atipicos por columna
def aplanar_con_pandas(lista_de_indices):
    return lista_de_indices[0].append(lista_de_indices[1:]).tolist()
valores_atipicos = aplanar_con_pandas(lista)

# eliminamos datos atipicos de df3
df3_sin_atipicos = df3.drop(valores_atipicos)
df3_sin_atipicos

# DESPUES: ya reemplazamos los valores NaN y liminamos los valores atipicos o extremos
df3_sin_atipicos.corr()
```
