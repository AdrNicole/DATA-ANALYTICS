```python
# 2. Crea una función que estime la proporción de hombres en el dataset, considerando 
# una muestra de 50 individuos, con niveles de confianza de 95% y 99%. Escribe con 
# palabras tus resultados al ejecutar la función e interpreta. Compara con la 
# proporción real. 


def estimar_proporcion_hombres(df, valor_hombre=1, n=50, random_state=42):
    # Proporción real (p) y muestral (p_hat)
    p_real = (df["male"] == valor_hombre).mean()
    p_hat = (df.sample(n=n, random_state=random_state)["male"] == valor_hombre).mean()

    # Error estándar y margen de error para 95% (z≈1.96) y 99% (z≈2.58)
    se = np.sqrt(p_hat * (1 - p_hat) / n)
    z_95, z_99 = stats.norm.ppf(0.975), stats.norm.ppf(0.995)

    # Construcción de la tabla de resultados
    res = pd.DataFrame({
        "Nivel": ["95%", "99%"],
        "Límite Inf": [round(max(0, p_hat - z * se), 4) for z in (z_95, z_99)],
        "p̂ Muestral": [round(p_hat, 4)] * 2,
        "Límite Sup": [round(min(1, p_hat + z * se), 4) for z in (z_95, z_99)],
    })

    print(f"Proporción Real (p): {p_real:.4f} ({p_real*100:.2f}%)")
    print(f"Proporción Muestral (p̂, n={n}): {p_hat:.4f}\n")
    print(res.to_string(index=False))

    return p_real, p_hat, res
# invocacion
estimar_proporcion_hombres(df)
```

```python
# 3. Considerando el dataset como una muestra, verifica si el género (male=1 significa 
# ‘hombre”) influye sobre el sueldo “earn” de las personas. Explica y justifica tu 
# procedimiento.  
```

```python
# Prueba de independencia entre el genero y el sueldo de las personas
import statsmodels.formula.api as smf
# 1. Separar sueldos por género
hombres = df[df["male"] == 1]["earn"].dropna()
mujeres = df[df["male"] == 0]["earn"].dropna()

# 2. Prueba t de Student (asumo muestras independientes)
t_stat, p_val = stats.ttest_ind(hombres, mujeres, equal_var=False)

# 3. Regresión lineal para medir la magnitud del impacto
modelo = smf.ols("earn ~ male", data=df).fit()

# Resultados
print(f"Sueldo Promedio Hombres: ${hombres.mean():,.2f}")
print(f"Sueldo Promedio Mujeres: ${mujeres.mean():,.2f}")
print(f"Diferencia observada:    ${(hombres.mean() - mujeres.mean()):,.2f}\n")
print(f"Estadístico t: {t_stat:.4f}")
print(f"p-valor:        {p_val:.4e}")

print("\n--- RESUMEN DEL MODELO DE REGRESIÓN ---")
print(modelo.summary().tables[1])
```
