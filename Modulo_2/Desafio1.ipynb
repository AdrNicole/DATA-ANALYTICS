
#Calculando el salario semanal


```python
salario_base = 500
comision = 0.1

monto_ventas = (int(input("Ingresar el monto de las ventas totales: ")))

# calculo de comision sobre salario
salario_total = salario_base + monto_ventas*comision
print("El saltrio total es: ",salario_total)
```

#Bienvenido a la frutería virtual






```python
catalogo_frutas = {'palta':2.8,
                   'fresa':10,
                   'pepinillo':9.8,
                   'Chirimoya':5.4}

# verificar si la fruta ingresada por usuario existe en catalogo
producto_usuario = input("Ingrese nombre de fruta: ")
if producto_usuario in catalogo_frutas:
  # mostrar el costo de la fruta si existe
  print(f'la fruta {producto_usuario} cuesta {catalogo_frutas[producto_usuario]}')

else:
  print('La fruta indicada no se encuentra en nuestra tienda')
```

#Bienvenido al casino

```python
import random

# escoger al azar 2 numeros entre 1 y 6
dado = [1,2,3,4,5,6]
E = random.sample(dado,2)

print('Resultado del experimento aleatorio de lanzar dos dados: ',E)

# sumar el numero de ambos elementos aleatorios
cont=0
for x in range(len(E)):
    cont+=E[x]
print('la suma de ambos valores es',cont)
```

#Calculadora de chocolate

```python
Persona_Kcal = float(input('Ingresar limite diario de Calorias: '))
Chocolate_Kcal = float(input('Ingresar la cantidad de calorias de la porcion de chocolate: '))

# Según Mapfre España, se aconseja que el consumo de azúcar represente como máximo el 10% de la ingesta calórica diaria, y a ser posible por debajo del 5%.

limite_diario_Azucar= Persona_Kcal*0.1

Porciones = limite_diario_Azucar//Chocolate_Kcal
print(f'La cantidad maxima de chocolate es de {Porciones} porciones')
```
